```python
def digit_sum(n):
    return sum(int(digit) for digit in str(n))
count = 0
for i in range(1, 1000): 
    if digit_sum(i) == 13:
        count += 1
print("Số lượng số nguyên dương nhỏ hơn 1000 có tổng các chữ số là 13:", count)
```
